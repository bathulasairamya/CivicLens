# CivicLens: 01. Data Exploration and Linguistic Diagnostics
This notebook explores the CivicLens complaint datasets, validates category distributions, analyzes script characteristics (Latin vs Indic scripts), and demonstrates leak-free grouped dataset partitioning.

```python
import os
import sys
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.preprocessing import clean_text, detect_script_characteristics, validate_complaints_data, grouped_train_test_split

sns.set_theme(style='whitegrid')
```

## 1. Load and Inspect Datasets

```python
df_complaints = pd.read_csv('../data/sample_complaints.csv')
print(f'Total complaints: {len(df_complaints)}')
df_complaints.head()
```

## 2. Category and Language Tag Distribution

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(data=df_complaints, y='category', order=df_complaints['category'].value_counts().index, ax=axes[0], palette='Blues_r')
axes[0].set_title('Complaints per Category', weight='bold')

sns.countplot(data=df_complaints, x='language_tag', order=df_complaints['language_tag'].value_counts().index, ax=axes[1], palette='Oranges_r')
axes[1].set_title('Linguistic Variations', weight='bold')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

## 3. Incident Grouping & Leak-Free Train-Test Partitioning

```python
train_df, test_df = grouped_train_test_split(df_complaints, group_col='incident_group_id', test_size=0.25, random_state=42)
print(f'Train size: {len(train_df)}, Test size: {len(test_df)}')

overlap = set(train_df['incident_group_id']).intersection(set(test_df['incident_group_id']))
print(f'Incident group leakage check: {len(overlap)} overlapping groups (Must be 0)')
assert len(overlap) == 0, 'Data leakage detected!'
```
